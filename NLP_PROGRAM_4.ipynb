```python
import re
text = input("Enter a string: ")
patterns = {
    "Starts with 'Hello' (^)": r"^Hello",
    "Ends with 'world' ($)": r"world$",
    "Contains a digit (\\d)": r"\d",
    "Contains a word (\\w)": r"\w",
    "Contains whitespace (\\s)": r"\s",
    "Contains 'cat'": r"cat",
    "Starts with a capital letter": r"^[A-Z]",
    "Ends with a digit": r"\d$"
}
for name, pattern in patterns.items():
    if re.search(pattern, text):
        print(name, ": YES")
    else:
        print(name, ": NO")
```
