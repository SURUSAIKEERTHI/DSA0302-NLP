```python
import re
patterns = {
    "Patient ID": r"\bPAT[0-9]{5}\b",
    "Date": r"\b(?:[0-2][0-9]|3[01])/(?:0[1-9]|1[0-2])/[0-9]{4}\b",
    "Drug Strength": r"\b[A-Za-z][A-Za-z0-9-]*\s+[0-9]+(?:\.[0-9]+)?\s*(?:mg|mcg|g|ml)\b",
    "Dosage Frequency": r"\b(?:OD|BD|TDS|QID|BID|once daily|twice daily)\b",
    "Laboratory Value": r"\b[A-Za-z][A-Za-z0-9-]*\s*[:=]?\s*[0-9]+(?:\.[0-9]+)?\s*(?:mg/dL|g/dL|mmol/L|IU/L|%)\b",
    "Blood Pressure": r"\b[0-9]{2,3}/[0-9]{2,3}\s*mmHg\b",
    "Temperature": r"\b[0-9]{2}(?:\.[0-9])?\s*°?[CF]\b",
    "ICD Code": r"\b[A-TV-Z][0-9][0-9A-TV-Z](?:\.[0-9A-TV-Z]{1,4})?\b",
    "Ward and Bed": r"\bWard\s+[A-Za-z0-9-]+\s*,?\s*Bed\s+[A-Za-z0-9-]+\b",
    "Contact Number": r"\b[6-9][0-9]{9}\b",
    "Age": r"\b(?:[1-9][0-9]?|1[01][0-9]|120)\s*(?:years?|yrs?)\b",
    "Symptom Duration": r"\bFever\s+duration\s*:\s*([0-9]+\s*(?:days?|weeks?|months?|years?|hours?))"
}
text = """
Patient ID: PAT12345
Admission Date: 06/11/2026
Prescription: Paracetamol 500mg
Frequency: BD
Glucose: 120 mg/dL
BP: 120/80 mmHg
Temperature: 37.5°C
Diagnosis: E11.9
Location: Ward A, Bed 12
Contact: 9876543210
Age: 45 years
Fever duration: 5 days
"""
actual = {
    "Patient ID": ["PAT12345"],
    "Date": ["06/11/2026"],
    "Drug Strength": ["Paracetamol 500mg"],
    "Dosage Frequency": ["BD"],
    "Laboratory Value": ["Glucose: 120 mg/dL"],
    "Blood Pressure": ["120/80 mmHg"],
    "Temperature": ["37.5°C"],
    "ICD Code": ["E11.9"],
    "Ward and Bed": ["Ward A, Bed 12"],
    "Contact Number": ["9876543210"],
    "Age": ["45 years"],
    "Symptom Duration": ["5 days"]
}
print("CLINICAL ENTITY EVALUATION")
print("-" * 85)
for entity, pattern in patterns.items():
    extracted = re.findall(pattern, text, re.IGNORECASE)
    if entity == "Symptom Duration":
        extracted = [match.strip() for match in extracted]
    expected = actual[entity]
    tp = len(set(extracted) & set(expected))
    fp = len(set(extracted) - set(expected))
    fn = len(set(expected) - set(extracted))
    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0
    f1 = (2 * precision * recall / (precision + recall)
          if precision + recall else 0)
    print(f"\nEntity: {entity}")
    print("TP:", tp, "FP:", fp, "FN:", fn)
    print("Precision:", round(precision, 2))
    print("Recall:", round(recall, 2))
    print("F1-Score:", round(f1, 2))
```

```python
import re
text = """
PAT12345 was admitted on 06/11/2026.
Prescription: Paracetamol 500MG.
Frequency: BD.
Glucose: 120 MG/DL.
BP: 120/80 MMHG.
Temperature: 37.5°C.
Diagnosis: E11.9.
Age: 45 YEARS.
Fever duration: 5 DAYS.
"""
normalized_text = text.lower()
normalized_text = re.sub(
    r"\b(\d{2})/(\d{2})/(\d{4})\b",
    r"\3-\2-\1",
    normalized_text
)
normalized_text = re.sub(r"\bmg\s*/\s*dl\b", "mg/dL", normalized_text)
normalized_text = re.sub(r"\bmmhg\b", "mmHg", normalized_text)
normalized_text = re.sub(r"\bmg\b", "mg", normalized_text)
abbreviations = {
    "od": "once daily",
    "bd": "twice daily",
    "tds": "three times daily",
    "qid": "four times daily",
    "bid": "twice daily"
}
for short_form, full_form in abbreviations.items():
    normalized_text = re.sub(
        r"\b" + short_form + r"\b",
        full_form,
        normalized_text,
        flags=re.IGNORECASE
    )
normalized_text = re.sub(r"[ \t]+", " ", normalized_text)
print("ORIGINAL CLINICAL TEXT")
print("----------------------")
print(text)
print("\nNORMALISED CLINICAL TEXT")
print("------------------------")
print(normalized_text.strip())
```

```python
def generate_plural(word):
    if word.endswith(("s", "sh", "ch", "x", "z")):
        return word + "es"
    elif word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ies"
    elif word.endswith("y"):
        return word + "s"
    else:
        return word + "s"
def generate_past_tense(word):
    if word.endswith("e"):
        return word + "d"
    elif word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ied"
    else:
        return word + "ed"
word = input("Enter a word: ").lower()
plural = generate_plural(word)
past = generate_past_tense(word)
print("\nORTHOGRAPHIC RULES")
print("------------------")
print("Root Word:", word)
print("Plural Form:", plural)
print("Past Tense Form:", past)
```

```python
import re
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
record = """
Patient ID: P1025
Age: 45 years
Admission Date: 12/10/2026
Discharge Date: 15/10/2026
Patient has fever for 5 days and cough.
Drug: Paracetamol 500mg
Dosage: 1 tablet BD
BP: 120/80 mmHg
Temperature: 38.5 C
Lab Value:  HbA1c 7.2 %
ICD Code: E11.9
Ward: W2 Bed: 15
Contact: 9876543210
"""
tokens = re.findall(
    r"\d+(?:\.\d+)?|[A-Za-z]+(?:[-'][A-Za-z]+)*",
    record
)
print("\n1. TOKENISATION")
print(tokens)
normalisation = {
    "bd": "twice daily",
    "od": "once daily",
    "tds": "three times daily",
    "qid": "four times daily"
}
normalised_tokens = [
    normalisation.get(token.lower(), token.lower())
    for token in tokens
]
print("\n2. TEXT NORMALISATION")
print(normalised_tokens)
patterns = {
    "Patient ID": r"Patient ID:\s*(P\d+)",
    "Age": r"Age:\s*(\d+\s*years?)",
    "Admission Date": r"Admission Date:\s*(\d{2}/\d{2}/\d{4})",
    "Discharge Date": r"Discharge Date:\s*(\d{2}/\d{2}/\d{4})",
    "Drug Name and Strength": r"Drug:\s*([A-Za-z]+\s*\d+\s*mg)",
    "Dosage Frequency": r"Dosage:\s*(.*)",
    "Blood Pressure": r"BP:\s*(\d{2,3}/\d{2,3}\s*mmHg)",
    "Temperature": r"Temperature:\s*(\d+(?:\.\d+)?\s*C)",
    "Lab Value": r"Lab Value:\s*([A-Za-z0-9]+\s+\d+(?:\.\d+)?\s*%)",
    "ICD Code": r"ICD Code:\s*([A-Z]\d{2}(?:\.\d+)?)",
    "Ward and Bed": r"Ward:\s*(\w+)\s*Bed:\s*(\d+)",
    "Contact Number": r"Contact:\s*([6-9]\d{9})",
    "Symptom Duration": r"(?:fever|cough)\s+for\s+(\d+\s+days?)"
}
print("\n3. ENTITY EXTRACTION")
for entity, pattern in patterns.items():
    match = re.search(pattern, record, re.IGNORECASE)
    if match:
        print(entity, ":", " ".join(match.groups()))
morphology = {
    "fractures": "fracture",
    "fractured": "fracture",
    "infections": "infection",
    "infected": "infect",
    "patients": "patient",
    "doctors": "doctor",
    "treated": "treat",
    "treating": "treat",
    "diagnoses": "diagnosis"
}
print("\n4. MORPHOLOGICAL ANALYSIS")
test_words = [
    "fractures", "infections", "patients",
    "doctors", "treated", "diagnoses"
]
for word in test_words:
    analysis = morphology.get(word, "Not found")
    print(word, "->", analysis)
print("\n5. PORTER STEMMING")
for word in test_words:
    print(word, "->", stemmer.stem(word))
print("\n6. PIPELINE COMPLETED")
print("Tokenisation: Completed")
print("Normalisation: Completed")
print("Entity Extraction: Completed")
print("Morphological Analysis: Completed")
print("Porter Stemming: Completed")
```
