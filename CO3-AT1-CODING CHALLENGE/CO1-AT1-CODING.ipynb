```python
import re
n = int(input())
lines = []
for _ in range(n):
    lines.append(input())
pattern = r"[A-Za-z]+(?:['-][A-Za-z]+)*"
print("\nInput:")
print(n)
for text in lines:
    print(text)
print("\nOutput:")
for text in lines:
    tokens = re.findall(pattern, text)
    if tokens:
        print(len(tokens), *tokens)
    else:
        print(0)
```

```python
n, k = map(int, input().split())
alphabet = input().split()
start = int(input())
data = list(map(int, input().split()))
m = data[0]
accepting = set(data[1:1 + m])
t = int(input())
transitions = {}
for _ in range(t):
    p, c, r = input().split()
    transitions.setdefault((int(p), c), set()).add(int(r))
Q = int(input())
queries = []
for _ in range(Q):
    queries.append(input().strip())
print("\nInput:")
print(n, k)
print(*alphabet)
print(start)
print(m, *sorted(accepting))
print(t)
for (p, c), destinations in transitions.items():
    for r in sorted(destinations):
        print(p, c, r)
print(Q)
for query in queries:
    print(query)
print("\nOutput:")
for s in queries:
    if s == "#":
        s = ""
    current = {start}
    for c in s:
        next_states = set()
        for state in current:
            next_states.update(transitions.get((state, c), set()))
        current = next_states
        if not current:
            break

    print("ACCEPT" if current & accepting else "REJECT")
```

```python
L = int(input())
lexicon = {}
for _ in range(L):
    stem, feature, form = input().split()
    lexicon[(stem, feature)] = form
Q = int(input())
queries = []
for _ in range(Q):
    stem, feature = input().split()
    queries.append((stem, feature))
def is_vowel(c):
    return c in "aeiou"
def is_doublable(stem):
    if len(stem) < 3:
        return False
    a, b, c = stem[-3], stem[-2], stem[-1]
    if is_vowel(a) or not is_vowel(b) or is_vowel(c):
        return False
    if c in "wxy":
        return False
    groups = 0
    inside = False
    for ch in stem:
        if is_vowel(ch):
            if not inside:
                groups += 1
            inside = True
        else:
            inside = False
    return groups == 1
def inflect(stem, feature):
    if (stem, feature) in lexicon:
        return lexicon[(stem, feature)]
    if feature in ("N+PL", "V+3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and not is_vowel(stem[-2]):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and not is_vowel(stem[-2]):
            return stem[:-1] + "ies"
        return stem + "s"
    if feature == "V+PROG":
        if len(stem) >= 3 and stem.endswith("e") and not is_vowel(stem[-2]):
            return stem[:-1] + "ing"
        if is_doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"
    suffixes = {
        "V+PAST": "ed",
        "ADJ+COMP": "er",
        "ADJ+SUP": "est"
    }
    if feature in suffixes:
        suffix = suffixes[feature]
        if stem.endswith("e"):
            if suffix == "ed":
                return stem + "d"
            if suffix == "er":
                return stem + "r"
            return stem + "st"
        if len(stem) >= 2 and stem.endswith("y") and not is_vowel(stem[-2]):
            return stem[:-1] + "i" + suffix
        if is_doublable(stem):
            return stem + stem[-1] + suffix
        return stem + suffix
    return stem
print("\nInput:")
print(L)
for (stem, feature), form in lexicon.items():
    print(stem, feature, form)
print(Q)
for stem, feature in queries:
    print(stem, feature)
print("\nOutput:")
for stem, feature in queries:
    print(inflect(stem, feature))
```

```python
def is_vowel(c):
    return c in "aeiou"
def is_doublable(stem):
    if len(stem) < 3:
        return False
    a, b, c = stem[-3], stem[-2], stem[-1]
    if is_vowel(a) or not is_vowel(b) or is_vowel(c):
        return False
    if c in "wxy":
        return False
    groups = 0
    inside = False
    for ch in stem:
        if is_vowel(ch):
            if not inside:
                groups += 1
            inside = True
        else:
            inside = False
    return groups == 1
def generate(stem, feature):
    if feature in ("PL", "3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and not is_vowel(stem[-2]):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and not is_vowel(stem[-2]):
            return stem[:-1] + "ies"
        return stem + "s"
    if feature == "PROG":
        if len(stem) >= 3 and stem.endswith("e") and not is_vowel(stem[-2]):
            return stem[:-1] + "ing"
        if is_doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"
    suffixes = {"PAST": "ed", "COMP": "er", "SUP": "est"}
    suffix = suffixes[feature]
    if stem.endswith("e"):
        if feature == "PAST":
            return stem + "d"
        if feature == "COMP":
            return stem + "r"
        return stem + "st"
    if len(stem) >= 2 and stem.endswith("y") and not is_vowel(stem[-2]):
        return stem[:-1] + "i" + suffix
    if is_doublable(stem):
        return stem + stem[-1] + suffix
    return stem + suffix
S = int(input())
lexicon = []
for _ in range(S):
    stem, pos = input().split()
    lexicon.append((stem, pos))
Q = int(input())
forms = []
for _ in range(Q):
    forms.append(input().strip())
print("\nInput:")
print(S)
for stem, pos in lexicon:
    print(stem, pos)
print(Q)
for form in forms:
    print(form)
print("\nOutput:")
for form in forms:
    analyses = set()
    for stem, pos in lexicon:
        if form == stem:
            analyses.add(stem + "+" + pos)
        if pos == "N":
            features = ["PL"]
        elif pos == "V":
            features = ["3SG", "PROG", "PAST"]
        else:
            features = ["COMP", "SUP"]
        for feature in features:
            if generate(stem, feature) == form:
                analyses.add(stem + "+" + pos + "+" + feature)
    print(", ".join(sorted(analyses)) if analyses else "UNKNOWN")
```

```python
def is_consonant(word, i):
    ch = word[i]
    if ch in "aeiou":
        return False
    if ch == 'y':
        if i == 0:
            return True
        return not is_consonant(word, i - 1)
    return True
def measure(word):
    m = 0
    i = 0
    length = len(word)
    while i < length and is_consonant(word, i):
        i += 1
    while i < length:
        while i < length and not is_consonant(word, i):
            i += 1
        if i < length:
            m += 1
            while i < length and is_consonant(word, i):
                i += 1
    return m
def has_vowel(word):
    return any(not is_consonant(word, i) for i in range(len(word)))
def double_consonant(word):
    return (
        len(word) >= 2
        and word[-1] == word[-2]
        and is_consonant(word, len(word) - 1)
    )
def cvc(word):
    if len(word) < 3:
        return False
    n = len(word)
    return (
        is_consonant(word, n - 3)
        and not is_consonant(word, n - 2)
        and is_consonant(word, n - 1)
        and word[-1] not in "wxy"
    )
def porter_step1(word):
    if word.endswith("sses"):
        word = word[:-2]
    elif word.endswith("ies"):
        word = word[:-2]
    elif word.endswith("ss"):
        pass
    elif word.endswith("s"):
        word = word[:-1]
    if word.endswith("eed"):
        stem = word[:-3]
        if measure(stem) > 0:
            word = stem + "ee"
    elif word.endswith("ed") and has_vowel(word[:-2]):
        word = word[:-2]
        word = cleanup(word)
    elif word.endswith("ing") and has_vowel(word[:-3]):
        word = word[:-3]
        word = cleanup(word)
    if word.endswith("y") and has_vowel(word[:-1]):
        word = word[:-1] + "i"
    return word
def cleanup(word):
    if word.endswith(("at", "bl", "iz")):
        return word + "e"
    if double_consonant(word) and word[-1] not in "lsz":
        return word[:-1]
    if measure(word) == 1 and cvc(word):
        return word + "e"
    return word
Q = int(input())
words = []
for _ in range(Q):
    words.append(input().strip())
print("\nInput:")
print(Q)
for word in words:
    print(word)
print("\nOutput:")
for word in words:
    print(porter_step1(word))
```
