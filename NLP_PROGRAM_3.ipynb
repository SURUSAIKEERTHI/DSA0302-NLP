```python
import re
text = input("Enter the text: ")
date = re.search(r'\b\d{2}/\d{2}/\d{4}\b', text)
print("First Date:", date.group() if date else "No date found")
mobiles = re.findall(r'\b[6-9]\d{9}\b', text)
print("Mobile Numbers:", mobiles)
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
print("Email IDs:", emails)
capitalized = re.findall(r'\b[A-Z]{2,}\b', text)
print("Capitalized Words:", capitalized)
if re.match(r'^CALL\b', text, re.IGNORECASE):
    print("Text starts with CALL")
else:
    print("Text does not start with CALL")
changed_date = re.sub(
    r'\b(\d{2})/(\d{2})/(\d{4})\b',
    r'\3-\2-\1',
    text
)
print("Changed Date Format:", changed_date)
hidden = re.sub(r'\b[6-9]\d{9}\b', 'XXXXXXXXXX', text)
print("Hidden Phone Numbers:", hidden)
```
