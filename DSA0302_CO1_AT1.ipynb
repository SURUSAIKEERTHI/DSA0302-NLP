```python
import re
def extract_name(resume):
    match = re.search(r"Name\s*:\s*([A-Za-z ]+)", resume, re.I)
    if match:
        return match.group(1).strip()
    return "Not Found"
def extract_emails(resume):
    return re.findall(r"[\w.-]+@[\w.-]+\.\w+", resume)
def extract_mobiles(resume):
    return re.findall(r"\b[6-9]\d{9}\b", resume)
def extract_skills(resume):
    skills = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
    found_skills = []
    for skill in skills:
        pattern = r"\b" + re.escape(skill) + r"\b"
        if re.search(pattern, resume, re.I):
            found_skills.append(skill)
    return found_skills
def extract_experience(resume):
    pattern = r"(\d+(?:\.\d+)?)\s*(?:years?|yrs?)\s*(?:of)?\s*experience"
    match = re.search(pattern, resume, re.I)
    if match:
        return float(match.group(1))
    return 0
def check_eligibility(experience, skills):
    return experience >= 2 and "Python" in skills
print("===== RESUME INFORMATION EXTRACTION SYSTEM =====")
resume = input("\nEnter resume text:\n")
name = extract_name(resume)
emails = extract_emails(resume)
mobiles = extract_mobiles(resume)
skills = extract_skills(resume)
experience = extract_experience(resume)
print("\n========== CANDIDATE PROFILE ==========")
print("Name       :", name)
if emails:
    print("Email      :", ", ".join(emails))
else:
    print("Email      : Not Found")

if mobiles:
    print("Mobile     :", ", ".join(mobiles))
else:
    print("Mobile     : Not Found")

if skills:
    print("Skills     :", ", ".join(skills))
else:
    print("Skills     : Not Found")
if experience > 0:
    print("Experience :", experience, "years")
else:
    print("Experience : Not Found")
print("\n========== ELIGIBILITY ==========")
if check_eligibility(experience, skills):
    print("Status     : ELIGIBLE")
    print("Reason     : Candidate has at least 2 years of experience")
    print("             and has Python skill.")
else:
    print("Status     : NOT ELIGIBLE")
    print("Reason     : Minimum requirement is 2 years of experience")
    print("             and Python skill.")
```

```python
import re
def search_products(products, keyword, search_type):
    matches = []
    if search_type == 1:
        pattern = r"^" + re.escape(keyword) + r"$"
    elif search_type == 2:
        pattern = r"^" + re.escape(keyword)
    elif search_type == 3:
        pattern = re.escape(keyword) + r"$"
    elif search_type == 4:
        pattern = re.escape(keyword)
    elif search_type == 5:
        pattern = re.escape(keyword)
    else:
        return matches
    for product in products:
        if search_type == 5:
            if re.search(pattern, product, re.IGNORECASE):
                matches.append(product)
        else:
            if re.search(pattern, product, re.IGNORECASE):
                matches.append(product)
    return matches
products = [
    "Apple iPhone 15",
    "Apple MacBook Air",
    "Apple AirPods",
    "Samsung Galaxy S24",
    "Samsung Galaxy Buds",
    "OnePlus 12",
    "OnePlus Nord CE 4",
    "Sony Headphones",
    "Dell Inspiron Laptop",
    "HP Laptop"
]
print("===== PRODUCT SEARCH SYSTEM =====")
print("\nAvailable Products:")
for product in products:
    print("-", product)
while True:
    print("\n========== SEARCH MENU ==========")
    print("1. Exact Keyword Search")
    print("2. Prefix Search")
    print("3. Suffix Search")
    print("4. Partial Keyword Search")
    print("5. Case-Insensitive Search")
    print("6. Exit")
    try:
        choice = int(input("\nEnter your choice: "))
        if choice == 6:
            print("\nSearch system closed.")
            break
        if choice < 1 or choice > 6:
            print("Invalid choice. Please try again.")
            continue
        keyword = input("Enter search keyword: ").strip()
        if keyword == "":
            print("Keyword cannot be empty.")
            continue
        matches = search_products(products, keyword, choice)
        print("\n========== SEARCH RESULTS ==========")
        if matches:
            print("Matching Products:")
            for product in matches:
                print("-", product)
            print("\nTotal Matching Products:", len(matches))
        else:
            print("No matching products found.")
            print("Total Matching Products: 0")
    except ValueError:
        print("Please enter a valid number.")
```

```python
import re
def validate(pattern, value):
    return re.fullmatch(pattern, value) is not None
print("===== UNIVERSITY REGISTRATION SYSTEM =====")
register_no = input("Enter Register Number: ").strip()
email = input("Enter Institutional Email: ").strip()
course_code = input("Enter Course Code: ").strip()
semester = input("Enter Semester: ").strip()
mobile = input("Enter Mobile Number: ").strip()
register_pattern = r"^[A-Za-z0-9]{5,15}$"
email_pattern = r"^[A-Za-z0-9._%+-]+@college\.edu$"
course_pattern = r"^[A-Za-z]{2,5}[0-9]{3}$"
semester_pattern = r"^[1-8]$"
mobile_pattern = r"^[6-9][0-9]{9}$"
register_valid = validate(register_pattern, register_no)
email_valid = validate(email_pattern, email)
course_valid = validate(course_pattern, course_code)
semester_valid = validate(semester_pattern, semester)
mobile_valid = validate(mobile_pattern, mobile)
print("\n========== VALIDATION RESULTS ==========")
if register_valid:
    print("Register Number : Valid")
else:
    print("Register Number : Invalid")
if email_valid:
    print("Institutional Email : Valid")
else:
    print("Institutional Email : Invalid")
if course_valid:
    print("Course Code : Valid")
else:
    print("Course Code : Invalid")
if semester_valid:
    print("Semester : Valid")
else:
    print("Semester : Invalid")
if mobile_valid:
    print("Mobile Number : Valid")
else:
    print("Mobile Number : Invalid")
print("\n========== FINAL REGISTRATION STATUS ==========")
if (register_valid and email_valid and course_valid
        and semester_valid and mobile_valid):
    print("Registration Successful")
else:
    print("Registration Failed")
    print("Please correct the invalid fields.")
```
