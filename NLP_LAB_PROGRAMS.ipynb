```python
import random
from collections import defaultdict
text = "I love NLP. I love Python. I love AI."
words = text.replace(".", "").split()
bigrams = defaultdict(list)
for i in range(len(words) - 1):
    bigrams[words[i]].append(words[i + 1])
word = "I"
result = [word]
for _ in range(6):
    if word not in bigrams:
        break
    word = random.choice(bigrams[word])
    result.append(word)
print("Generated Text:", " ".join(result))
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
text = "The smart student learns Python"
tokens = nltk.word_tokenize(text)
tags = nltk.pos_tag(tokens)
print("Tokens:", tokens)
print("POS Tags:", tags)
```

```python
words = ["I", "love", "Python"]

tag_probabilities = {
    "I": {"PRP": 0.95, "NN": 0.05},
    "love": {"VBP": 0.90, "NN": 0.10},
    "Python": {"NNP": 0.85, "NN": 0.15}
}

print("Stochastic POS Tagging:")

for word in words:
    best_tag = max(
        tag_probabilities[word],
        key=tag_probabilities[word].get
    )
    print(word, "->", best_tag)
```

```python
import re
patterns = [
    (r".*ing$", "VBG"),
    (r".*ed$", "VBD"),
    (r".*ly$", "RB"),
    (r".*ous$", "JJ"),
    (r".*", "NN")
]
words = ["playing", "walked", "quickly", "famous", "book"]
print("Rule-Based POS Tagging:")
for word in words:
    for pattern, tag in patterns:
        if re.fullmatch(pattern, word):
            print(word, "->", tag)
            break
```

```python
words = ["They", "can", "fish"]
tags = ["PRP", "MD", "NN"]
print("Before Transformation:")
print(list(zip(words, tags)))
for i in range(len(words) - 1):
    if words[i] == "can" and words[i + 1] == "fish":
        tags[i] = "VB"
print("\nAfter Transformation:")
print(list(zip(words, tags)))
```

```python
import nltk
grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'boy' | 'apple'
V -> 'eats'
""")
parser = nltk.RecursiveDescentParser(grammar)
sentence = "the boy eats the apple".split()
print("Parse Tree:")
for tree in parser.parse(sentence):
    print(tree)
```

```python
import nltk
grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'girl' | 'ball'
V -> 'kicks'
""")
parser = nltk.ChartParser(grammar)
sentence = "the girl kicks the ball".split()
print("Parse Tree:")
for tree in parser.parse(sentence):
    tree.pretty_print()
```

```python
import nltk
grammar = nltk.CFG.fromstring("""
S -> SINGULAR SV
S -> PLURAL PV
SINGULAR -> 'he' | 'she'
PLURAL -> 'they' | 'we'
SV -> 'runs' | 'eats'
PV -> 'run' | 'eat'
""")
parser = nltk.ChartParser(grammar)
sentences = [
    "he runs",
    "they run",
    "he run",
    "they runs"
]
for sentence in sentences:
    tokens = sentence.split()
    trees = list(parser.parse(tokens))
    if trees:
        print(sentence, "-> Correct")
    else:
        print(sentence, "-> Incorrect")
```

```python
import nltk
grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> ProperNoun [0.5] | CommonNoun [0.5]
VP -> V NP [0.7] | V [0.3]
ProperNoun -> 'John' [0.5] | 'Mary' [0.5]
CommonNoun -> 'apples' [0.6] | 'mangoes' [0.4]
V -> 'likes' [1.0]
""")
parser = nltk.ViterbiParser(grammar)
sentence = "John likes apples".split()
print("PCFG Parse Tree:")
for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```
