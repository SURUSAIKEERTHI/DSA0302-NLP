```python
import re
text = input("Enter a text: ")
phone = re.search(r'\d{10}', text)
email = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)
if phone:
    print("Phone number:", phone.group())
else:
    print("Phone number not found")

if email:
    print("Email:", email)
else:
    print("Email not found")
```

```python
def fsa(string):
    state = 'q0'
    for ch in string:
        if state == 'q0':
            if ch == 'a':
                state = 'q1'
            else:
                state = 'q0'
        elif state == 'q1':
            if ch == 'b':
                state = 'q2'
            elif ch == 'a':
                state = 'q1'
            else:
                state = 'q0'
        elif state == 'q2':
            if ch == 'a':
                state = 'q1'
            else:
                state = 'q0'
    return state == 'q2'
string = input("Enter a string: ")
if fsa(string):
    print("Accepted")
else:
    print("Rejected")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
word = input("Enter a word: ")
lemmatizer = WordNetLemmatizer()
root = lemmatizer.lemmatize(word)
print("Original word:", word)
print("Root word:", root)
```

```python
def plural(noun):
    if noun.endswith(("s", "sh", "ch", "x", "o")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    else:
        return noun + "s"
word = input("Enter a noun: ")
print("Plural form:", plural(word))
```
